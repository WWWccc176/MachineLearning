```python
### Exercise 1: Plot sine and cosine functions on the same figure
###
### Task:
### Create one plot that shows both sin(x) and cos(x)
### for x values ranging from -pi to pi using Numpy and Matplotlib.


### Requirements: 
### 1. Generate x values from -pi to pi.
### 2. Compute y1 = sin(x) and y2 = cos(x).
### 3. Use plt.plot() to plot both curves on the same figure.
### 4. Use different colors for the two curves.
### 5. Use different line styles for the two curves.
### 6. Add x-axis label.
### 7. Add y-axis label.
### 8. Add a title.
### 9. Add a legend.
### 10. Add grid lines.
### 11. Display the figure using plt.show().
### 12. Save the figure to your computer with name 'practice1.png'


```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 生成 -pi 到 pi 的 x 值
x = np.linspace(-np.pi, np.pi, 300)

# 2. 计算正弦和余弦值
y1 = np.sin(x)
y2 = np.cos(x)

# 3. 在同一张图上绘制两条曲线（颜色、线型不同）
plt.plot(x, y1, color='blue', linestyle='-', label='sin(x)')
plt.plot(x, y2, color='red', linestyle='--', label='cos(x)')

# 6. x 轴标签
plt.xlabel('x')

# 7. y 轴标签
plt.ylabel('y')

# 8. 标题
plt.title('Sine and Cosine Functions')

# 9. 图例
plt.legend()

# 10. 网格线
plt.grid(True)

# 11. 显示图像
plt.show()

# 12. 保存为 practice1.png（放在当前工作目录）
plt.savefig('practice1.png')
```
